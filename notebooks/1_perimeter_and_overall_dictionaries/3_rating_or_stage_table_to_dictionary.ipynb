## Create a JSON from an excel table ##

```python
import os
import re
import pathlib as pl
import pandas as pd
import json
```

```python
#set working directory and folder variables
os.chdir('..')
os.chdir('..')
home = pl.Path(os.getcwd())
```

```python
home = pl.Path(home)
print('home is at ',home)
```

```python
#inputs for this project
name_of_excel = 'huc_outlet_type.xlsx'
project = 'wy_fy23'
```

```python
#set up input and output folders
inputs = home/'inputs'
outputs_base = home/'outputs'
#create input and outputs folder if not already dirs
if not os.path.exists(inputs/project/'rating_stage_table'):
    os.makedirs(inputs/project/'rating_stage_table')
if not os.path.exists(outputs_base/project/'event_tie_ins'):
    os.makedirs(outputs_base/project/'event_tie_ins')
```

```python
try:
     df = pd.read_excel(inputs/project/'rating_stage_table'/name_of_excel, sheet_name='Sheet1')
except ValueError as e:
    print(f"Error reading sheet 'Sheet1': {e}")
    exit()

#convert excel table to json file for use in event tie in creation
json_dict = dict(zip(df['huc'], df['bc']))


#create json file from dictionary
json_file = outputs_base/project/'event_tie_ins'/'rating_or_stage.json'
with open(json_file, 'w') as f:
        json.dump(json_dict, f, indent=4)
```
